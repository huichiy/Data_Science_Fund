## 1. Introduction & Problem Statement

### 1.1 Main Objective

The central objective of this analysis is to investigate how economic development and energy consumption patterns influence renewable energy adoption and carbon emissions globally between 2000 and 2020. This research aims to produce evidence-based conclusions using the data science methodology approach on a complete data set from across countries. It is aimed at providing useful insights that may assist in designing sustainable energy policies and hence advance the progress towards achieving Sustainable Development Goal (SDG) 7.

### 1.2 Background and Context

One of the major challenges of the twenty-first century involves the shift from traditional energy systems to sustainable energy systems. The generation of energy accounts for the leading cause of greenhouse gas emissions by humans. Decarbonization of energy systems plays an important role in achieving the commitments made in the 2030 Agenda for Sustainable Development (United Nations, 2015). Even though there have been substantial developments in the use of renewable sources over the past two decades, there remain many differences across countries in relation to electricity access, share of renewables, and carbon intensity. It is important to gain an understanding of the factors behind these differences.

Apergis and Payne (2010) demonstrated a bidirectional causal relationship between renewable energy consumption and economic growth across OECD countries, suggesting that the two processes are mutually reinforcing over the long run. In contrast, Owusu and Asumadu-Sarkodie (2016) identified institutional and financial barriers as primary constraints on renewable adoption in lower-income countries, irrespective of technological progress. Furthermore, the International Energy Agency (IEA, 2021) documents persistent heterogeneity in energy profiles across income levels, with high-income nations exhibiting higher absolute renewable capacity but not necessarily higher renewable shares, partly due to the classification of traditional biomass as renewable in international accounting frameworks. This observation is further supported by Ritchie and Roser (2022), who show that electricity access and per capita energy consumption are strongly positively correlated with GDP per capita, thereby complicating straightforward comparisons of renewable share across income groups.

### 1.3 Problem Statement

> **How do economic development and energy consumption patterns influence renewable energy adoption and carbon emissions globally between 2000 and 2020?**

### 1.4 Analytical Questions

Four distinct analytical questions are derived from the problem statement, each employing a different methodological approach:

| Question | Description | Technique |
|----------|-------------|-----------|
| **Q1** | What is the relationship between a country's GDP per capita and its renewable energy share in total final energy consumption? | Pearson & Spearman Correlation |
| **Q2** | To what extent do renewable energy capacity and energy intensity predict CO₂ emissions across countries over time? | Multiple Linear Regression |
| **Q3** | Can countries be grouped into meaningful clusters based on their energy access, renewable share, and carbon emissions profiles? | K-Means Clustering |
| **Q4** | Can a country's level of electricity access be classified as high or low based on its economic and energy indicators? | Random Forest Classification |

### 1.5 Dataset Suitability

The *Global Data on Sustainable Energy (2000–2020)* dataset is particularly well-suited for this investigation for three reasons. Firstly, its multi-country, multi-year panel structure (176 countries × 21 years) enables robust cross-national comparisons and temporal trend analysis. Secondly, it contains variables that directly operationalise the constructs of interest — GDP per capita, renewable share, CO₂ emissions, electricity access and energy intensity — thereby facilitating all four analytical questions without requiring external data merging. Lastly, the dataset originates from recognised international institutions (World Bank, IEA, Our World in Data), thereby ensuring methodological consistency and reliability across country-year observations.

### Limitations

- **High missingness:** `Financial flows to developing countries` (57% missing) and `Renewables (% equivalent primary energy)` (59% missing) have limited analytical utility and will be handled pairwise.
- **Traditional biomass classification:** Countries with limited electricity grid access often report high renewable shares due to reliance on traditional biomass (wood, charcoal), which is classified as renewable under international standards. This may inflate renewable share figures for lower-income countries.
- **National-level aggregation:** The dataset operates at country level; sub-national variation is not captured.
- **Coverage gaps:** Not all 176 countries have complete data for all 21 years, introducing potential reporting bias for smaller and lower-income nations.
- **Column naming:** One column (`Density
(P/Km2)`) contains an embedded newline character, which must be corrected programmatically before analysis.

## 4. Data Cleaning & Preprocessing

All cleaning steps are applied to a copy of `df_raw`; the original in-memory dataset is never modified.

### 4.1 Create Working Copy

A copy of `df_raw` is created so that the original in-memory dataset remains unchanged. This ensures that any step can be rerun from the raw data without needing to reload from file.

```python
df = df_raw.copy()
print(f"Working copy created: {df.shape}")
```

### 4.2 Standardise Column Names

**Justification:** The raw column `Density
(P/Km2)` contains an embedded newline character, which causes indexing errors. Additionally, mixed casing, special characters, and measurement units embedded in column names reduce code readability. All columns are therefore renamed to consistent snake_case identifiers.

```python
rename_map = {
    'Entity'                                                          : 'country',
    'Year'                                                            : 'year',
    'Access to electricity (% of population)'                         : 'access_electricity_pct',
    'Access to clean fuels for cooking'                               : 'access_clean_fuels_pct',
    'Renewable-electricity-generating-capacity-per-capita'            : 'renewable_capacity_per_capita',
    'Financial flows to developing countries (US $)'                  : 'financial_flows_usd',
    'Renewable energy share in the total final energy consumption (%)': 'renewable_share_pct',
    'Electricity from fossil fuels (TWh)'                             : 'electricity_fossil_twh',
    'Electricity from nuclear (TWh)'                                  : 'electricity_nuclear_twh',
    'Electricity from renewables (TWh)'                               : 'electricity_renewable_twh',
    'Low-carbon electricity (% electricity)'                          : 'low_carbon_electricity_pct',
    'Primary energy consumption per capita (kWh/person)'              : 'energy_consumption_per_capita',
    'Energy intensity level of primary energy (MJ/$2017 PPP GDP)'     : 'energy_intensity',
    'Value_co2_emissions_kt_by_country'                               : 'co2_emissions_kt',
    'Renewables (% equivalent primary energy)'                        : 'renewables_pct_primary',
    'gdp_growth'                                                      : 'gdp_growth',
    'gdp_per_capita'                                                  : 'gdp_per_capita',
    r'Density\n(P/Km2)'                                               : 'population_density',
    'Land Area(Km2)'                                                  : 'land_area_km2',
    'Latitude'                                                        : 'latitude',
    'Longitude'                                                       : 'longitude'
}

df.rename(columns=rename_map, inplace=True)
print("Column names standardised.")
print(df.columns.tolist())
```

### 4.3 Fix Data Types

`year` is stored as an integer and `country` as a string to ensure correct sorting and grouping behaviour. `population_density` loads as text because dense countries use comma thousands-separators (e.g. `8,358`); the commas are stripped and the column is converted to a numeric type so it can be used in calculations.

```python
df['year']    = df['year'].astype(int)
df['country'] = df['country'].astype(str)
df['population_density'] = pd.to_numeric(
    df['population_density'].astype(str).str.replace(',', '', regex=False),
    errors='coerce'
)

print(df[['country', 'year', 'population_density']].dtypes)
```

### 4.4 Remove Duplicate Rows

**Justification:** Exact duplicate rows would inflate sample size and bias all statistical estimates, including correlation coefficients and regression parameters.

```python
n_before = len(df)
df.drop_duplicates(inplace=True)
print(f"Rows before: {n_before:,}  |  Rows after: {len(df):,}  |  Removed: {n_before - len(df)}")
```

### 4.5 Handle Missing Values

**Strategy:** Retain all rows; handle missing values pairwise at the point of each analysis (complete-case analysis).

**Justification:** Dropping rows globally would discard valid data for variables with low missingness. Columns with >40% missing (`financial_flows_usd`, `renewables_pct_primary`) are flagged but retained for potential use. The only rows dropped globally are those missing both `country` and `year`, as these are entirely uninformative.

```python
high_missing = df.isnull().mean()
high_missing = high_missing[high_missing > 0.40].sort_values(ascending=False)
print("Columns with >40% missing (flagged, retained):")
print(high_missing.apply(lambda x: f"{x*100:.1f}%").to_string())

before = len(df)
df.dropna(subset=['country', 'year'], inplace=True)
print(f"Rows dropped (missing country/year): {before - len(df)}")
```

### 4.6 Outlier Inspection

**Strategy:** Outliers are identified using the IQR method but are **retained**.

**Justification:** Extreme values in this dataset represent genuine country-level variation (e.g., Qatar's exceptionally high GDP per capita, Norway's high renewable share, China's high CO₂ emissions). Removing them would distort global cross-national comparisons, which are the central focus of this analysis.

```python
key_cols = ['gdp_per_capita', 'renewable_share_pct',
            'co2_emissions_kt', 'energy_consumption_per_capita',
            'access_electricity_pct']

print(f"  {'Column':<35} {'IQR-Low':>10} {'IQR-High':>10} {'# Outliers':>12}")
print("  " + "-" * 70)
for col in key_cols:
    d    = df[col].dropna()
    Q1   = d.quantile(0.25)
    Q3   = d.quantile(0.75)
    IQR  = Q3 - Q1
    lo, hi = Q1 - 1.5*IQR, Q3 + 1.5*IQR
    n_out  = ((d < lo) | (d > hi)).sum()
    print(f"  {col:<35} {lo:>10.1f} {hi:>10.1f} {n_out:>12}")

print("All outliers retained.")
```

## 5. Exploratory Data Analysis (EDA)
This EDA section uses the cleaned dataset from 2000 to 2020.  

The main goals are:

- Examine the distribution of CO₂ emissions.
- Understand why CO₂ emissions may require log transformation.
- Explore relationships between numerical variables.
- Understand the panel structure, since each country appears across multiple years.

- Check the electricity access distribution for the classification task.

> **Note on labels:** Throughout this EDA, **Q1, Q2, Q3, and Q4** refer to the four analytical questions defined in Section 1.4.


```python
# Create derived columns for per-capita analysis
# Estimate population using population density and land area.
# This is used only for EDA visualisation, not as an official population figure.
df_eda = df.copy()
df_eda['population_est'] = df_eda['population_density'] * df_eda['land_area_km2']

# Convert total CO₂ emissions from kilotonnes to tonnes per person.
df_eda['co2_per_capita'] = df_eda['co2_emissions_kt'] * 1_000 / df_eda['population_est']  

print("EDA frame:", df_eda.shape)

# Quick check: the largest estimated populations should look realistic.
print("\nTop 10 derived populations (2020):")
print(df_eda[df_eda['year'] == 2020]
      .sort_values('population_est', ascending=False)
      [['country', 'population_est']].head(10).to_string(index=False))
```

### 5.1 Univariate distributions
Distributions of the core economic and energy indicators. Shapes here justify
transforms used later (e.g. the heavy right-skew of GDP and CO2).

```python
uni_cols = {
    'gdp_per_capita'               : 'GDP per Capita (USD)',
    'renewable_share_pct'          : 'Renewable Share (%)',
    'access_electricity_pct'       : 'Electricity Access (%)',
    'energy_consumption_per_capita': 'Energy Consumption/cap (kWh)',
    'energy_intensity'             : 'Energy Intensity (MJ/$ PPP)',
    'low_carbon_electricity_pct'   : 'Low-Carbon Electricity (%)',
}

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
for ax, (col, label) in zip(axes.ravel(), uni_cols.items()):
    sns.histplot(df_eda[col].dropna(), bins=40, ax=ax, color='#2D6A4F')
    ax.set_title(label)
    ax.set_xlabel('')
fig.suptitle('Univariate Distributions of Key Indicators',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 5.2 CO₂ emissions — raw vs log
Total CO2 is dominated by a few large emitters (China, USA), producing extreme
right-skew. The log view is far more symmetric, which **motivates a log-scale /
per-capita target for the Q2 regression** rather than modeling raw kt.

```python
co2 = df_eda['co2_emissions_kt'].dropna()
co2 = co2[co2 > 0]

fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
sns.histplot(co2, bins=45, ax=ax[0], color='#B5432A')
ax[0].set_title('CO$_2$ emissions (kt) — raw, right-skewed')
ax[0].set_xlabel('CO$_2$ emissions (kt)')

sns.histplot(np.log10(co2), bins=45, ax=ax[1], color='#55A868')
ax[1].set_title('log$_{10}$ CO$_2$ emissions — approx. symmetric')
ax[1].set_xlabel('log$_{10}$ CO$_2$ emissions (kt)')
plt.tight_layout()
plt.show()
```

### 5.3 Top 10 CO₂-emitting countries (2019)
Drilling into the CO₂ skew shown in 5.2: a handful of countries account for the
bulk of global emissions. This concentration is *why* raw CO₂ in kilotonnes is
a poor regression target for Q2 — a few countries would dominate the fit and
the model would mostly "learn" their identity rather than the underlying drivers.
The log / per-capita target in Q2 normalises this. Using 2019 because 2020 CO₂
values are missing from the dataset.

```python
year = 2019
top = (df_eda[df_eda['year'] == year]
       .dropna(subset=['co2_emissions_kt'])
       .nlargest(10, 'co2_emissions_kt')
       .sort_values('co2_emissions_kt'))   # ascending for horizontal bar

fig, ax = plt.subplots(figsize=(9, 4.8))
sns.barplot(y='country', x='co2_emissions_kt', data=top,
            color='#B5432A', ax=ax)
ax.set_title(f'Top 10 CO$_2$-emitting countries ({year})')
ax.set_xlabel('CO$_2$ emissions (kt)'); ax.set_ylabel('')
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x/1e6:.1f}M'))
plt.tight_layout()
plt.show()

global_y = df_eda[df_eda['year'] == year]['co2_emissions_kt'].sum()
top5_share = top.tail(5)['co2_emissions_kt'].sum() / global_y * 100
print(f"Top 5 emitters = {top5_share:.1f}% of {year} global emissions  "
      f"(China alone {top['co2_emissions_kt'].max()/global_y*100:.1f}%).")
```

### 5.4 Global CO₂ emissions trend (2000–2019)
Aggregating across all countries shows the overall trajectory: emissions rose
by roughly 50% over the period, with a small visible dip around 2008–2009
(the global financial crisis). 2020 is omitted because the dataset has no CO₂
values for that year. The trend is steady enough that pooling 2000–2019 rows
for modelling is reasonable, but each country still has its own trajectory —
which is why country-level splits remain essential for Q2 and Q4.

```python
trend = (df_eda.dropna(subset=['co2_emissions_kt'])
         .groupby('year')['co2_emissions_kt'].sum().reset_index())

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(trend['year'], trend['co2_emissions_kt'] / 1e6,
        color='#B5432A', lw=2.2, marker='o', ms=5)
ax.set_title('Global CO$_2$ emissions, 2000–2019')
ax.set_xlabel('Year'); ax.set_ylabel('Total CO$_2$ emissions (Gt)')
ax.grid(True, alpha=0.3)
ax.axvspan(2008, 2009, color='grey', alpha=0.12, label='2008–09 GFC')
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()

g0 = trend['co2_emissions_kt'].iloc[0] / 1e6
g1 = trend['co2_emissions_kt'].iloc[-1] / 1e6
print(f"Global CO$_2$: {g0:.1f} Gt ({trend['year'].iloc[0]}) → "
      f"{g1:.1f} Gt ({trend['year'].iloc[-1]})  ({(g1/g0-1)*100:+.0f}%).")
```

### 5.5 Global electricity mix trend (Fossil / Renewable / Nuclear)
The three electricity-source columns are mechanically additive — they sum
(approximately) to total generation — so they are perfectly collinear when used
together as predictors. This plot is the visual version of the leakage point
the all-numeric correlation heatmap (5.8) will quantify: in Q2 and Q4 these
three columns must not be used jointly as features. The trend itself shows
fossil generation still dominating but renewables growing fastest, while
nuclear has stayed roughly flat over two decades.

```python
sources = {'electricity_fossil_twh': 'Fossil',
           'electricity_renewable_twh': 'Renewable',
           'electricity_nuclear_twh': 'Nuclear'}
mix = df_eda.groupby('year')[list(sources.keys())].sum().reset_index()

fig, ax = plt.subplots(figsize=(9, 4.5))
colors = {'Fossil': '#B5432A', 'Renewable': '#2D6A4F', 'Nuclear': '#4C72B0'}
for col, label in sources.items():
    ax.plot(mix['year'], mix[col], marker='o', ms=4, lw=2,
            color=colors[label], label=label)
ax.set_title('Global electricity generation by source, 2000–2020')
ax.set_xlabel('Year'); ax.set_ylabel('Generation (TWh)')
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

deltas = ((mix.iloc[-1, 1:] / mix.iloc[0, 1:] - 1) * 100).round(0)
print("Change 2000 → 2020:",
      ", ".join(f"{sources[c]} {d:+.0f}%" for c, d in deltas.items()))
```

### 5.7 Bivariate relationships (motivating Q2–Q4)
Pairwise views that preview the modeling questions other than Q1 (Q1's own

### Q1 — Correlation Analysis *(Member 1)*

**Question:** What is the relationship between a country's GDP per capita and its renewable energy share in total final energy consumption?

**Technique:** Pearson correlation (linear association), Spearman rank correlation (monotonic association), scatter plot with OLS trend line, Pearson correlation heatmap.

**Justification:** Pearson *r* measures the strength of a linear association between two continuous variables. Spearman *ρ* captures monotonic relationships and is more robust to outliers and non-normality. Employing both provides a richer characterisation of the GDP–renewable energy nexus, particularly given the heavily right-skewed distribution of GDP per capita.

#### 8.1.1 Data Preparation

```python
q1_df = df[['country', 'year', 'gdp_per_capita', 'renewable_share_pct']].dropna()

print(f"Observations : {len(q1_df):,}")
print(f"Countries    : {q1_df['country'].nunique()}")
print(f"Year range   : {q1_df['year'].min()} – {q1_df['year'].max()}")
display(q1_df[['gdp_per_capita', 'renewable_share_pct']].describe().round(2))
```

#### 8.1.2 Scatter Plot — GDP per Capita vs Renewable Energy Share

A log scale is applied to the x-axis because GDP per capita is heavily right-skewed. The OLS trend line is fitted on log-transformed GDP values to linearise the relationship for visual clarity.

```python
fig, ax = plt.subplots(figsize=(9, 5.5))

ax.scatter(
    q1_df['gdp_per_capita'],
    q1_df['renewable_share_pct'],
    alpha=0.22, s=13, color='#2D6A4F', edgecolors='none',
    label='Country-year observation'
)

valid  = q1_df[q1_df['gdp_per_capita'] > 0].copy()
slope, intercept, *_ = stats.linregress(
    np.log10(valid['gdp_per_capita']),
    valid['renewable_share_pct']
)
x_range = np.linspace(valid['gdp_per_capita'].min(), valid['gdp_per_capita'].max(), 400)
ax.plot(x_range, slope * np.log10(x_range) + intercept,
        color='#B5432A', linewidth=2.2, label='OLS trend (log GDP scale)')

ax.set_xscale('log')
ax.set_xlabel('GDP per Capita (USD, log scale)', fontsize=11)
ax.set_ylabel('Renewable Energy Share (%)', fontsize=11)
ax.set_title(
    'Q1 — GDP per Capita vs Renewable Energy Share\n'
    'in Total Final Energy Consumption (2000–2020)',
    fontsize=13, fontweight='bold', pad=12
)
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'\${x:,.0f}'))
ax.legend(fontsize=9, framealpha=0.8)
ax.grid(True, alpha=0.25, linestyle='--')

plt.tight_layout()
plt.show()
```

#### 8.1.3 Pearson & Spearman Correlation Tests

Both tests are two-tailed. Statistical significance is assessed at α = 0.05. Correlation strength is interpreted using the conventional |r| thresholds: < 0.20 very weak, 0.20–0.39 weak, 0.40–0.59 moderate, 0.60–0.79 strong, ≥ 0.80 very strong.


```python
x = q1_df['gdp_per_capita']
y = q1_df['renewable_share_pct']

pearson_r,  pearson_p  = stats.pearsonr(x, y)
spearman_r, spearman_p = stats.spearmanr(x, y)
n = len(q1_df)

def strength(r):
    r = abs(r)
    if r < 0.20: return "Very weak"
    if r < 0.40: return "Weak"
    if r < 0.60: return "Moderate"
    if r < 0.80: return "Strong"
    return "Very strong"

results = pd.DataFrame({
    'Metric'       : ['Pearson r', 'Spearman ρ'],
    'Coefficient'  : [round(pearson_r, 4), round(spearman_r, 4)],
    'p-value'      : [f'{pearson_p:.3e}', f'{spearman_p:.3e}'],
    'Direction'    : ['Negative', 'Negative'],
    'Strength'     : [strength(pearson_r), strength(spearman_r)],
    'Significant?' : ['Yes (p < 0.0001)', 'Yes (p < 0.0001)']
})

print(f"n = {n:,}")
display(results)
```

### Q2 — Multiple Linear Regression *(Member 2)*

**Question:** To what extent do renewable energy capacity and energy intensity predict CO₂ emissions across countries over time?

**Technique:** Multiple Linear Regression (MLR) with standardised predictors and a country-level train/test split.

**Why this technique:** The MLR estimates the strength of association between multiple numerical predictors and a numerical outcome variable. It generates (i) an estimate of a coefficient for each predictor – which can be interpreted as "change in log₁₀ CO₂ per standard deviation change in predictor variable, other variables held constant" – and (ii) an R² statistic that indicates the proportion of total variance accounted for.


#### 8.2.1 Data Preparation, Target Choice, and Train/Test Split

**Target choice — `log₁₀(CO₂ per capita)`.**  
EDA Section 5.2 showed that raw CO₂ in kilotonnes is heavily right-skewed and dominated by country *size* (China, USA), not by energy policy. Using per-capita CO₂ removes the size effect, and a log transform symmetrises the distribution — together they let MLR satisfy its linearity / homoscedasticity assumptions.

**Predictors.**  
Two focal predictors (the ones the question asks about): `renewable_capacity_per_capita` and `energy_intensity`. Two controls (so the focal coefficients are not biased by wealth and energy demand): `log₁₀(gdp_per_capita)` and `energy_consumption_per_capita`.

**Excluded by design (leakage flagged in Section 5.11):** `electricity_fossil_twh`, `electricity_renewable_twh`, `electricity_nuclear_twh`, `low_carbon_electricity_pct`, `renewable_share_pct` — these are mechanically tied to the CO₂ outcome.

**Train/Test split.**  
`GroupShuffleSplit` by `country` (80/20) — required by Section 5.11 so the same country never appears in both train and test. `StandardScaler` is fitted on the training set only.


```python
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# Per-capita CO₂ (tonnes/person) requires an estimated population (density × land area)
q2_df = df.copy()
q2_df['population_est']    = q2_df['population_density'] * q2_df['land_area_km2']
q2_df['co2_per_capita_t']  = q2_df['co2_emissions_kt'] * 1_000 / q2_df['population_est']

q2_cols = ['country', 'year', 'co2_per_capita_t',
           'renewable_capacity_per_capita', 'energy_intensity',
           'gdp_per_capita', 'energy_consumption_per_capita']
q2_df = q2_df[q2_cols].dropna()
q2_df = q2_df[(q2_df['co2_per_capita_t'] > 0) & (q2_df['gdp_per_capita'] > 0)]

q2_df['log_co2_pc'] = np.log10(q2_df['co2_per_capita_t'])
q2_df['log_gdp']    = np.log10(q2_df['gdp_per_capita'])

feature_cols = ['renewable_capacity_per_capita', 'energy_intensity',
                'log_gdp', 'energy_consumption_per_capita']
X      = q2_df[feature_cols].values
y      = q2_df['log_co2_pc'].values
groups = q2_df['country'].values

print(f"Observations : {len(q2_df):,}")
print(f"Countries    : {q2_df['country'].nunique()}")
print(f"Year range   : {q2_df['year'].min()} – {q2_df['year'].max()}")

# Country-level 80/20 split — no country appears in both train and test
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
g_train, g_test = groups[train_idx], groups[test_idx]

# Fit scaler on TRAIN ONLY
scaler  = StandardScaler().fit(X_train)
X_train_s = scaler.transform(X_train)
X_test_s  = scaler.transform(X_test)

print(f"\nTrain  : {len(X_train):>5,} rows  |  {len(set(g_train)):>3} countries")
print(f"Test   : {len(X_test):>5,} rows  |  {len(set(g_test)):>3} countries")
print(f"Country overlap (must be 0): {len(set(g_train) & set(g_test))}")
```

#### 8.2.2 Fit MLR and Inspect Coefficients

The standardised coefficient (β) reports the change in `log₁₀(CO₂ per capita)` per **one standard deviation** change in each predictor, holding the others constant. Because all predictors are on the same standardised scale, |β| values are directly comparable and reveal which driver matters most.


```python
model = LinearRegression().fit(X_train_s, y_train)

coef_df = pd.DataFrame({
    'Feature'             : feature_cols,
    'Std. coefficient (β)': model.coef_.round(4),
    'Direction'           : ['↓ CO₂' if c < 0 else '↑ CO₂' for c in model.coef_],
})
coef_df = coef_df.reindex(coef_df['Std. coefficient (β)'].abs().sort_values(ascending=False).index)
print(f"Intercept (mean log₁₀ CO₂ per capita on train, standardised inputs): {model.intercept_:.4f}")
display(coef_df.reset_index(drop=True))
```

#### 8.2.3 Evaluation Metrics

- **R²** — proportion of variance in `log₁₀(CO₂ per capita)` explained by the model
- **RMSE / MAE** — average prediction error in log₁₀ units (an RMSE of 0.30 means typical prediction is within ≈10^0.30 ≈ 2× of actual)
- We report both train and test metrics. A small train-test gap indicates the model generalises to unseen countries.


```python
y_pred_train = model.predict(X_train_s)
y_pred_test  = model.predict(X_test_s)

metrics = pd.DataFrame({
    'Set'  : ['Train', 'Test'],
    'R²'   : [r2_score(y_train, y_pred_train),
              r2_score(y_test,  y_pred_test)],
    'RMSE' : [np.sqrt(mean_squared_error(y_train, y_pred_train)),
              np.sqrt(mean_squared_error(y_test,  y_pred_test))],
    'MAE'  : [mean_absolute_error(y_train, y_pred_train),
              mean_absolute_error(y_test,  y_pred_test)],
}).round(4)
display(metrics)
```

#### 8.2.4 Diagnostic Plots

Three side-by-side diagnostics:
1. **Coefficient bar chart** — visual ranking of effect sizes (sign + magnitude)
2. **Predicted vs Actual** (test set) — points on the 45° line indicate accurate predictions
3. **Residuals vs Predicted** — should look like random noise around zero; any pattern violates linearity / homoscedasticity


```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4.6))

# (1) Coefficient bar chart
labels = {
    'renewable_capacity_per_capita': 'Renewable\ncapacity/cap',
    'energy_intensity'             : 'Energy\nintensity',
    'log_gdp'                      : 'log GDP\nper capita',
    'energy_consumption_per_capita': 'Energy\nconsumption/cap',
}
coefs = pd.Series(model.coef_, index=[labels[f] for f in feature_cols]).sort_values()
colors = ['#2D6A4F' if v < 0 else '#B5432A' for v in coefs]
axes[0].barh(coefs.index, coefs.values, color=colors)
axes[0].axvline(0, color='black', lw=0.8)
axes[0].set_title('Standardised coefficients\n(MLR — log₁₀ CO₂ per capita)', fontweight='bold')
axes[0].set_xlabel('β (per 1 SD predictor)')

# (2) Predicted vs Actual (test set)
axes[1].scatter(y_test, y_pred_test, alpha=0.35, s=18,
                color='#4C72B0', edgecolors='none')
lims = [min(y_test.min(), y_pred_test.min()), max(y_test.max(), y_pred_test.max())]
axes[1].plot(lims, lims, 'r--', lw=1.5, label='45° (perfect fit)')
axes[1].set_xlabel('Actual log₁₀ CO₂ per capita')
axes[1].set_ylabel('Predicted log₁₀ CO₂ per capita')
axes[1].set_title('Predicted vs Actual (test set)', fontweight='bold')
axes[1].legend()
axes[1].grid(True, alpha=0.25)

# (3) Residuals vs Predicted
residuals = y_test - y_pred_test
axes[2].scatter(y_pred_test, residuals, alpha=0.35, s=18,
                color='#55A868', edgecolors='none')
axes[2].axhline(0, color='red', lw=1.2, linestyle='--')
axes[2].set_xlabel('Predicted log₁₀ CO₂ per capita')
axes[2].set_ylabel('Residual (actual − predicted)')
axes[2].set_title('Residuals vs Predicted', fontweight='bold')
axes[2].grid(True, alpha=0.25)

plt.tight_layout()
plt.show()
```

### Q3 — K-Means Clustering *(Member 3)*

**Question:** Can countries be grouped into meaningful clusters based on their energy access, renewable share, and carbon emissions profiles?

**Technique:** K-Means clustering with standardised features, K chosen via the elbow method (WCSS) and silhouette score, and visualised through PCA projection and a cluster-profile heatmap.

**Why this technique:** The K-means clustering algorithm is the classic example of an unsupervised partitioning technique — that is, it partitions countries into K clusters, in which each country is most similar to the mean characteristics of its cluster. The method generates (i) easily interpretable cluster centers from which we can *name* each category and (ii) a set number of categories that relate well to the policy story.


#### 8.3.1 Data Preparation and Standardisation

**Unit of analysis — one row per country.** Each country gets a single profile, taken from the **2019 snapshot** (the most recent year with complete CO₂ data — Section 5.4 noted 2020 CO₂ is missing). If we kept all 21 country-years, the same country could land in different clusters across years, which is meaningless.

**Features (taken literally from the question):**

| Feature | Source column | Captures |
|---|---|---|
| Electricity access | `access_electricity_pct` | Grid coverage |
| Clean fuels access | `access_clean_fuels_pct` | Cooking-fuel modernisation |
| Renewable share | `renewable_share_pct` | Renewables in final energy |
| Carbon emissions | `log₁₀(co2_per_capita_t)` | Emissions normalised by population, log-transformed (Section 5.2) |

`StandardScaler` is applied because K-Means is **distance-based** and would otherwise be dominated by the feature with the largest raw range.

**Outliers:** Section 4.6 retained outliers as genuine variation. Section 5.11 warned that Q3 must avoid singleton clusters — the log-transform on CO₂ already compresses the extreme range, so retention is safe. We verify cluster sizes after fitting.


```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# 2019 snapshot — one row per country
q3_df = df.copy()
q3_df['population_est']   = q3_df['population_density'] * q3_df['land_area_km2']
q3_df['co2_per_capita_t'] = q3_df['co2_emissions_kt'] * 1_000 / q3_df['population_est']

q3_snap = q3_df[q3_df['year'] == 2019][[
    'country',
    'access_electricity_pct', 'access_clean_fuels_pct',
    'renewable_share_pct', 'co2_per_capita_t'
]].dropna()
q3_snap = q3_snap[q3_snap['co2_per_capita_t'] > 0].copy()
q3_snap['log_co2_pc'] = np.log10(q3_snap['co2_per_capita_t'])

q3_features = ['access_electricity_pct', 'access_clean_fuels_pct',
               'renewable_share_pct', 'log_co2_pc']

print(f"Snapshot year   : 2019")
print(f"Countries       : {len(q3_snap)}")
display(q3_snap[q3_features].describe().round(2))

scaler_q3 = StandardScaler().fit(q3_snap[q3_features])
X_q3 = scaler_q3.transform(q3_snap[q3_features])
```

#### 8.3.2 Selecting K — Elbow and Silhouette

Two diagnostics for choosing K:
- **WCSS / elbow method** — within-cluster sum of squares; lower is better but always decreases with K. The "elbow" (point of diminishing returns) suggests a good K.
- **Silhouette score** — measures how well each point fits its cluster vs. neighbouring clusters. Range is [-1, 1]; higher is better. This is the **primary criterion**.

We sweep K = 2 … 8 and report both.


```python
k_range = list(range(2, 9))
wcss, sil = [], []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X_q3)
    wcss.append(km.inertia_)
    sil.append(silhouette_score(X_q3, km.labels_))

k_table = pd.DataFrame({'K': k_range, 'WCSS': np.round(wcss, 1), 'Silhouette': np.round(sil, 3)})
display(k_table)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

axes[0].plot(k_range, wcss, 'o-', color='#4C72B0', lw=2, ms=8)
axes[0].set_xlabel('Number of clusters (K)')
axes[0].set_ylabel('WCSS (inertia)')
axes[0].set_title('Elbow method', fontweight='bold')
axes[0].grid(True, alpha=0.3)

axes[1].plot(k_range, sil, 'o-', color='#B5432A', lw=2, ms=8)
axes[1].axvline(3, color='grey', linestyle='--', alpha=0.6, label='Chosen K = 3')
axes[1].set_xlabel('Number of clusters (K)')
axes[1].set_ylabel('Silhouette score')
axes[1].set_title('Silhouette analysis', fontweight='bold')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

**K = 3 chosen.** Although K = 2 has a marginally higher silhouette (0.59 vs 0.52), it produces a trivial "developed vs developing" split that adds no analytical value. K = 3 reveals a meaningful *transitioning mid-income* tier between the two extremes, and no cluster is smaller than 26 countries (no singletons — Section 5.11 constraint satisfied).

The cluster centroid table below shows the mean profile of each group. Cluster names are assigned based on these profiles.


```python
K_FINAL = 3
kmeans = KMeans(n_clusters=K_FINAL, random_state=42, n_init=10).fit(X_q3)
q3_snap = q3_snap.copy()
q3_snap['cluster'] = kmeans.labels_

# Cluster centroid profiles (means on original scale)
profile = q3_snap.groupby('cluster')[q3_features].mean().round(2)
profile['co2_per_capita_t'] = q3_snap.groupby('cluster')['co2_per_capita_t'].mean().round(2)
profile['n_countries']      = q3_snap.groupby('cluster').size()

# Assign descriptive names based on observed profiles
cluster_names = {
    0: 'Low-access biomass-dependent',
    1: 'Transitioning mid-income',
    2: 'Industrialised high-emission',
}
profile.index = [f"C{i}: {cluster_names[i]}" for i in profile.index]

print("Cluster centroid profiles (2019 means, original units):")
display(profile)

# Verify no singleton clusters
print(f"\nMin cluster size: {profile['n_countries'].min()}  (singleton-cluster check: PASS)")
print(f"Silhouette at K={K_FINAL}: {silhouette_score(X_q3, kmeans.labels_):.3f}")
```

### Q4 — Classification Analysis *(Member 4)*

**Question:** Can a country's level of electricity access be classified as high or low based on its economic and energy indicators?

**Technique:** Random Forest Classification

**Why this technique:** We use a Random Forest because it is an ensemble method that combines many weak predictors into one strong predictor. It is like asking a group of experts to vote on the answer, giving us a more accurate and stable prediction than any single model alone.


#### 8.4.1 Data Preparation

First, we need to change our target variable (`access_electricity_pct`) into two categories. We will define "High Access" as 90% or above (Class 1), and "Low Access" as below 90% (Class 0). Then, we split the data into a training set (80%) to teach the model, and a testing set (20%) to check its performance.

```python
from sklearn.model_selection import train_test_split
from sklearn.model_selection import GroupShuffleSplit
q4_df = df[['access_electricity_pct', 'gdp_per_capita', 'renewable_share_pct',
            'energy_consumption_per_capita', 'country']].dropna()

q4_df['access_class'] = (q4_df['access_electricity_pct'] >= 90.0).astype(int)

print("Class distribution:")
print(q4_df['access_class'].value_counts())
print(f"Imbalance ratio: {q4_df['access_class'].value_counts().min() / q4_df['access_class'].value_counts().max():.2f}\n")

feature_cols = ['gdp_per_capita', 'renewable_share_pct', 'energy_consumption_per_capita']
X = q4_df[feature_cols]
y = q4_df['access_class']
groups = q4_df['country']

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))
X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

print(f"Total observations used : {len(q4_df):,}")
print(f"Training data size      : {len(X_train):,}")
print(f"Testing data size       : {len(X_test):,}")
print(f"\nTraining countries : {groups.iloc[train_idx].nunique()}")
print(f"Testing countries  : {groups.iloc[test_idx].nunique()}")
print("(No country appears in both sets)")
```

#### 8.4.2 Model Training and Evaluation

Now, we instantiate a `RandomForestClassifier` with 200 estimators, applying `class_weight='balanced'` to handle class imbalance. To guard against overfitting, tree complexity is constrained by setting `max_depth=15` and `min_samples_leaf=5`. The model is then fitted to the training set, utilizing Out-of-Bag (OOB) estimation alongside a 5-fold `GroupKFold` cross-validation to verify predictive stability.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix


rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=5,
    class_weight='balanced',
    oob_score=True,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)

print("--- Random Forest Classification Results ---")
print(f"Accuracy Score: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Low Access (<90%)', 'High Access (>=90%)']))

from sklearn.model_selection import cross_val_score, GroupKFold

print(f"OOB Score     : {rf_model.oob_score_:.4f}\n")

gkf = GroupKFold(n_splits=5)
cv_scores = cross_val_score(rf_model, X, y, groups=groups,
                             cv=gkf, scoring='f1_macro')
print(f"5-Fold GroupKFold CV — F1 (macro): {cv_scores.mean():.4f} ± {cv_scores.std():.4f}")
```

#### 8.4.3 Feature Importance Visualization

To understand how the model works and check its accuracy, we look at three plots. The bar chart shows which indicator has the biggest impact on a country's electricity access. At the same time, the confusion matrix and ROC curve show that the model can easily tell high-access countries apart from low-access ones.

```python
feature_importances = pd.DataFrame({
    'Feature': ['GDP per Capita', 'Renewable Share (%)', 'Energy Consumption per Capita'],
    'Importance': rf_model.feature_importances_
}).sort_values(by='Importance', ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(x='Importance', y='Feature', data=feature_importances, palette='viridis', ax=ax)

ax.set_title('Feature Importance for Predicting Electricity Access', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Importance Score', fontsize=12)
ax.set_ylabel('')
plt.tight_layout()
plt.show()

from sklearn.metrics import roc_auc_score, RocCurveDisplay

y_pred_proba = rf_model.predict_proba(X_test)[:, 1]
roc_auc = roc_auc_score(y_test, y_pred_proba)
print(f"ROC-AUC Score : {roc_auc:.4f}\n")

cm = confusion_matrix(y_test, y_pred)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Low (<90%)', 'High (≥90%)'],
            yticklabels=['Low (<90%)', 'High (≥90%)'],
            ax=axes[0])
axes[0].set_title('Confusion Matrix', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Predicted')
axes[0].set_ylabel('Actual')

RocCurveDisplay.from_predictions(y_test, y_pred_proba, ax=axes[1])
axes[1].set_title(f'ROC Curve (AUC = {roc_auc:.4f})', fontsize=14, fontweight='bold')
axes[1].plot([0, 1], [0, 1], 'k--', lw=1)

plt.tight_layout()
plt.show()
```

## 9. Results & Interpretation

### Q1 — Correlation Analysis

The Pearson correlation coefficient between GDP per capita and renewable energy share is *r* = **−0.371** (*p* < 0.0001, *n* = 3,189), indicating a statistically significant weak-to-moderate negative linear relationship. Furthermore, the Spearman rank correlation yields *ρ* = **−0.651** (*p* < 0.0001), a notably stronger coefficient that reflects a moderate-to-strong monotonic but non-linear relationship. Both results are statistically significant at all conventional significance levels.

The stronger Spearman coefficient relative to Pearson confirms that the relationship is monotonic but non-linear — a log-transformation of GDP per capita linearises the pattern, as shown in the scatter plot above.

**Heatmap findings:**
- The strongest negative correlation in the dataset is between renewable share and electricity access (*r* = −0.785). Countries with limited electricity grid access rely disproportionately on traditional biomass, which is classified as renewable, thereby inflating their renewable share figure.
- GDP per capita correlates most strongly with energy consumption per capita (*r* = +0.667) and electricity access (*r* = +0.418).
- Energy intensity is negatively associated with GDP per capita (*r* = −0.185), consistent with the finding that wealthier economies are more energy-efficient per unit of output.

**Interpretation:** The negative direction of the correlation does not indicate that wealthier countries avoid renewables. Rather, it reflects a compositional effect — lower-income countries report higher renewable shares because of biomass reliance, whereas wealthier nations have diversified, fossil-fuel-intensive energy infrastructures that dilute the renewable share proportion despite large absolute investments in clean energy. These findings are consistent with Owusu and Asumadu-Sarkodie (2016) and the IEA (2021).

The MLR model achieves **test-set R² ≈ 0.81** (RMSE ≈ 0.31 in log₁₀ units, MAE ≈ 0.24), indicating that roughly 81% of the cross-country variation in log₁₀(CO₂ per capita) is captured by just four economic and energy indicators — and that performance generalises to *entirely unseen countries* (no country leaks across the 80/20 split).

**Interpretation:**

2. **Energy intensity matters in the expected direction.** Countries that need *more* energy per dollar of GDP emit more per person — confirming that efficiency gains (lower intensity) are a meaningful lever for decarbonisation independent of wealth.

3. **Renewable capacity has a small negative effect.** Higher installed renewable capacity per capita is associated with slightly *lower* per-capita CO₂, but the effect is modest. A single global "increase renewable share" target is insufficient because the metric is not comparable across clusters.

## 10. Conclusions & Limitations

### 10.1 Synthesis of Findings

### 10.2 Overall Insight

The four analyses converge on a single, robust insight: **economic development is by far the strongest force shaping global CO₂ emissions, while renewable share — taken in isolation — is a misleading metric for cross-country comparison**. High renewable shares in low-income countries reflect biomass dependence (Owusu & Asumadu-Sarkodie, 2016; IEA, 2021), not modern renewable infrastructure. Meaningful decarbonisation therefore requires a *segmented* policy response: industrialised high-emission countries (Cluster 2) need efficiency-focused interventions targeting energy intensity, while low-access countries (Cluster 0) need energy-justice interventions that displace biomass with modern clean energy without locking in fossil dependence. The cross-country evidence on green energy practices reported by Yu et al. (2021) supports this segmented framing.

### 10.3 Limitations

1. **Biomass classification.** International accounting (Ritchie & Roser, 2022) classifies traditional biomass as renewable. This inflates renewable-share metrics for low-income countries and is the root cause of the negative Q1 correlation. A future analysis should split "modern renewables" from "traditional biomass".
2. **Country-level aggregation.** All findings are at the national level. Within-country variation (urban vs rural, sub-national energy mix) is invisible to this analysis.
3. **High missingness in two variables.** `Financial flows to developing countries` (57% missing) and `Renewables (% equivalent primary energy)` (59% missing) were excluded from analysis, limiting the financial dimension of the investigation.
4. **2020 CO₂ data missing.** Q3 used a 2019 snapshot rather than the most recent year, so any pandemic-era shifts are not captured.
5. **Linear functional form in Q2.** MLR assumes a linear log-CO₂ relationship — possible non-linear interactions (e.g. renewable effectiveness conditional on grid quality) are not modelled.
6. **Causality.** All four analyses describe *associations*, not causal effects. Apergis and Payne (2010) document a bidirectional renewable–GDP relationship that linear models cannot resolve.

### 10.4 Implications and Future Work

- **Policy implication.** A uniform global "increase renewable share" target is insufficient because the metric is not comparable across country profiles. Sustainable Development Goal 7 monitoring (United Nations, 2015) would benefit from cluster-specific sub-targets.
- **Methodological extension.** Replace MLR with panel fixed-effects regression or gradient-boosted regression (XGBoost) to capture within-country dynamics and non-linear interactions.
- **Data extension.** Merge with sub-national grid-quality data (e.g. Open Power System Data) to test whether renewable capacity becomes a stronger emissions-reduction lever when grid integration is high.
- **Causal inference.** Apply instrumental-variable or difference-in-differences designs around specific renewable-energy policy adoptions (feed-in tariffs, carbon pricing) to test causality.
- **Sectoral analysis.** Extend to firm-level data on sustainable supply chain practices (Onukwulu et al., 2021) to bridge the gap between national emissions targets and operational decarbonisation.
